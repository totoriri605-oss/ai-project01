```python
customers = [
    {"name": "민수", "amount": 49000, "grade": "일반"},
    {"name": "지영", "amount": 50000, "grade": "VIP"},
    {"name": "현우", "amount": 99000, "grade": "일반"},
    {"name": "수빈", "amount": 100000, "grade": "VIP"},
    {"name": "원빈", "amount": 120000, "grade": "일반"},
]

for customer in customers:
    name = customer["customer"]
    amount = customer["amount"]
    grade = customer["grade"]
    ship_type = " "
    if amount >= 100000 and grade == "VIP":
       ship_type = "VIP 우선배송"
    elif amount >= 50000 or grade == "VIP":
      ship_type = "무료배송"
    else:
      ship_type = "일반배송"

    print(f"{name}님의 배송 결과는 {ship_type} 입니다. 주문금액 : {amount}원, 회원등급 : {grade}")
```

```python
import csv

products = [
    {"name": "키보드", "rating": 4.8, "stock": 12, "status": "판매중"},
    {"name": "마우스", "rating": 4.6, "stock": 3, "status": "판매중"},
    {"name": "모니터", "rating": 4.2, "stock": 0, "status": "판매중"},
    {"name": "헤드셋", "rating": 4.9, "stock": 8, "status": "판매중지"},
    {"name": "웹캠", "rating": 3.8, "stock": 10, "status": "판매중"}
]

results = []

for product in products:
    name = product["name"]
    rating = product["rating"]
    stock = product["stock"]
    status = product["status"]

    if status != "판매중" or stock <= 0:
        result = "판매불가"

    elif rating >= 4.5 and stock >= 5:
        result = "추천"

    elif rating >= 4.0:
        result = "일반"

    else:
        result = "관심 필요"

    print(name, rating, stock, status, result)

    results.append({
        "상품명": name,
        "평점": rating,
        "재고": stock,
        "판매상태": status,
        "결과": result
    })


with open(
    "products_result.csv",
    "w",
    newline="",
    encoding="utf-8-sig"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=["상품명", "평점", "재고", "판매상태", "결과"]
    )

    writer.writeheader()
    writer.writerows(results)

print("CSV 저장 완료!")
```
