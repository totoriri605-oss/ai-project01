### 가장 어려웠던 문제와 이유

회원 구매액 판정 문제가 가장 어려웠다.
List 전체 이름과 한 개의 데이터 이름을 구분하는 부분,
Dictionary에서 값을 꺼내는 부분,
for문과 if/else의 들여쓰기를 맞추는 부분이 헷갈렸다.

```python
students = [
    {"name": "민수", "score": 95, "attendance": 90},
    {"name": "지영", "score": 85, "attendance": 95},
    {"name": "현우", "score": 72, "attendance": 65},
    {"name": "수진", "score": 78, "attendance": 88}

]

for student in students:
    name = student["name"]
    score = student["score"]
    attendance = student["attendance"]

    if score < 95 or attendance < 69:

        grade = "F"
    elif score >= 90 and attendance >= 80:
        grade = "A"
    elif score >= 80 and attendance >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    else:
        grade = "D"

    print(

        f"{name}: 점수 {score}점, "
        f"출석률 {attendance}%, 등급 {grade}"
    )
```

### 반복
- 무엇을 하나씩 확인하는가? 학생 한명 씩 확인한다.

### 조건
- F가 되는 조건: 점수가 60점 미만이거나 출석률이 70 미만일 때
- A가 되는 조건: 점수가 90점 이상이고 출석률이 80 이상일 때
- B가 되는 조건: 점수가 80점 이상이고 출석률이 80 이상일 때
- C가 되는 조건: 점수가 70점 이상일 때
- D가 되는 조건: 위 조건에 모두 해당하지 않을 때

### 새로운 표현
- `and`를 사용한 곳: A와 B 등급 조건에서 점수와 출석률을 둘 다 만족하는지 확인할 때
- `or`를 사용한 곳: F 등급 조건에서 점수 또는 출석률 중 하나라도 기준보다 낮은지 확인할 때
- `elif`가 필요한 이유를 지금 생각해 본 내용: 여러 조건 중 하나에 해당하는 등급을 차례대로 확인하기 위해 사용한다.
- f-string에서 `{name}`의 역할을 지금 생각해 본 내용: `name` 변수에 들어 있는 학생 이름을 문장 안에 넣어 출력한다.
